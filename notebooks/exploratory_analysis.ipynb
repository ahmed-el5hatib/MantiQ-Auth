## 1. Load and Visualize Sample Images

```python
# Find available test images
data_dir = Path('../data/dicom_processed')
images = sorted(data_dir.rglob('*.png'))[:6]

if images:
    fig, axes = plt.subplots(2, 3, figsize=(12, 8))
    for ax, img_path in zip(axes.flat, images):
        img = plt.imread(str(img_path))
        ax.imshow(img, cmap='gray')
        ax.set_title(img_path.stem[:20])
        ax.axis('off')
    plt.suptitle('Sample LIDC-IDRI Images', fontsize=14)
    plt.tight_layout()
    plt.show()
else:
    print('No processed images found. Run preprocessing first.')
```

## 2. Feature Extraction Analysis

```python
# Compare ViT and ResNet features on sample image
if images:
    img_path = images[0]
    tensor = load_and_preprocess_image(img_path)
    
    vit_features = extract_vit_features(tensor)
    resnet_features = extract_resnet_features(tensor)
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 4))
    axes[0].bar(range(len(vit_features)), vit_features, width=1.0)
    axes[0].set_title(f'ViT-B/16 Features ({len(vit_features)}-dim)')
    axes[0].set_xlabel('Dimension')
    
    axes[1].bar(range(len(resnet_features)), resnet_features, width=1.0)
    axes[1].set_title(f'ResNet-18 Features ({len(resnet_features)}-dim)')
    axes[1].set_xlabel('Dimension')
    
    plt.tight_layout()
    plt.show()
```
